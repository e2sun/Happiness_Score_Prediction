```python
#Importing necessary libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.preprocessing import StandardScaler
import numpy as np
```

```python
# Prepare data
df.fillna(df.mean(numeric_only=True), inplace=True)
df = df.clip(lower=df.quantile(0.01), upper=df.quantile(0.99), axis=1)
df.rename(columns={'Life Ladder': 'Happiness_Score', 'Log GDP per capita': 'Log_GDP'}, inplace=True)
if 'Country' in df.columns:
    df = pd.get_dummies(df, columns=['Country'], drop_first=True)
corr = df.corr()['Happiness_Score'].abs().sort_values(ascending=False)
relevant_features = corr[corr > 0.1].index
df = df[relevant_features]
```

```python
# Fit model
X = df.drop("Happiness_Score", axis=1)
y = df["Happiness_Score"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=12)
model = LinearRegression()

model.fit(X_train, y_train)
lin_pred= model.predict(X_test)
```

```python
# Evaluating model
lin_mse = mean_squared_error(y_test, lin_pred)
lin_rmse = np.sqrt(lin_mse)
lin_mae = mean_absolute_error(y_test, lin_pred)
r2 = r2_score(y_test, lin_pred)

print(f"Mean Absolute Error (MAE): {mae:.4f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")
print(f"R² Score: {r2:.4f}")
```

## Commentary on Actual vs. Predicted Plots

The Actual vs. Predicted plots provide visual insight into how well each model captures the true values of the target variable (`Happiness_Score`):

### Linear Regression
- The points are loosely scattered around the red dashed line (perfect prediction line), indicating that the model captures some general trends but struggles with more complex patterns.  
- The spread suggests underfitting, especially where actual happiness scores are higher or lower than average.  
- This is expected from a linear model, which assumes a straight-line relationship between features and the target.

### Random Forest
- The points are much more tightly clustered around the red dashed line, especially across the full range of actual happiness scores.  
- This indicates that the Random Forest model is able to capture nonlinear relationships and better fit the data.  
- Predictions are more accurate and consistent across different value ranges, showing less bias and lower variance in error compared to Linear Regression.

### Summary
The Random Forest model shows superior performance in predicting happiness scores, both visually and quantitatively. The tighter clustering around the ideal prediction line demonstrates that it is able to model complex interactions between features more effectively than Linear Regression.
